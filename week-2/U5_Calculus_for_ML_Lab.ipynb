```python
import numpy as np
import sympy as sp

x, y = sp.symbols('x y')
sp.init_printing()
np.random.seed(42)
print('Setup complete. SymPy', sp.__version__, '| NumPy', np.__version__)
```

```python
# NUMERICAL DERIVATIVE (finite difference)
```

```python
def f(x):
    return x**2
h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')
```

```python
# SYMBOLIC DERIVATIVE (SymPy)
```

```python
expr = x**2
deriv = sp.diff(expr, x)
print('d/dx (x**2) =', deriv)
print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

```python
# PARTIAL DERIVATIVES
```

```python
f2 = x**2 + 3*x*y + y**2
print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```

```python
# THE GRADIENT (vector of partials)
```

```python
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)
grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)
```

```python
# CHAIN RULE BY HAND vs SymPy
```

```python
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)
print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

```python
# CHAINING THREE FUNCTIONS
```

```python
expr3 = (3*x + 1)**4
print('d/dx (3x+1)^4 =', sp.diff(expr3, x))
```

```python
# FORWARD PASS
```

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1
h  = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

```python
# BACKWARD PASS (gradients via the chain rule)
```

```python
dy   = 2 * (y_hat - Y) / Y.size
dW2  = h.T @ dy
dh   = dy @ W2.T
dz1  = dh * (z1 > 0)
dW1  = X.T @ dz1
print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')
```

```python
# ONE GRADIENT-DESCENT STEP SHOULD LOWER THE LOSS
```

```python
lr = 0.1
W1 -= lr * dW1
W2 -= lr * dW2
h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

```python
# THE HESSIAN (matrix of second derivatives = curvature)
```

```python
f5 = x**2 + 3*x*y + y**2
H = sp.hessian(f5, (x, y))
print('Hessian of f:')
sp.pprint(H)
```

```python
# GRADIENT DESCENT ON A SIMPLE FUNCTION
```

```python
xv = 0.0
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)
    xv = xv - lr * grad
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```

```python
import sympy as sp
# 1. Hessian of x**4 + y**2
x, y = sp.symbols('x y')
f = x**4 + y**2
H = sp.hessian(f, (x, y))
print("Hessian matrix:")
sp.pprint(H)
```

```python
# 2. Gradient descent to minimise (x - 7)**2
xv = 0.0
lr = 0.1

for step in range(20):
    grad = 2 * (xv - 7)
    xv = xv - lr * grad

print('Final x:', xv)
```
